```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

df = pd.read_csv('titanic_clean.csv')
df.sample(4)
```

```python
# 1. Distribución de edades
plt.figure(figsize=(10,6))
sns.histplot(df, x='Age', kde=True)
plt.title('Distribución de Edades')
plt.axvline(df['Age'].mean(), color='red')
plt.axvline(df['Age'].median(), color='green')
plt.show()

# 2. Sobrevivientes por clase
plt.figure(figsize=(10,6))
sns.countplot(df, x='Pclass', hue='Survived')
plt.title('Sobrevivientes por Clase')
plt.show()

# 3. Sobrevivientes por género
plt.figure(figsize=(10,6))
sns.countplot(df, x='Sex', hue='Survived')
plt.title('Sobrevivientes por Género')
plt.show()

# 4. Sobrevivientes por puerto de embarque
plt.figure(figsize=(10,6))
sns.countplot(df, x='Embarked', hue='Survived')
plt.title('Sobrevivientes por Puerto de Embarque')
plt.show()

# Matriz de correlación
plt.figure(figsize=(10,6))
sns.heatmap(df[['Survived','Pclass','Age','SibSp','Parch','Fare']].corr(), annot=True, cmap='coolwarm')
plt.title('Correlación de variables')
plt.show()
```
