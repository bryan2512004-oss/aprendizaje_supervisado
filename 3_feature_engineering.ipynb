```python
import pandas as pd
from sklearn.preprocessing import LabelEncoder, MinMaxScaler, QuantileTransformer

df = pd.read_csv('titanic_clean.csv')

# Convertir texto a números
label_sex = LabelEncoder()
label_embarked = LabelEncoder()
df['Sex'] = label_sex.fit_transform(df['Sex'])
df['Embarked'] = label_embarked.fit_transform(df['Embarked'])

# Normalizar Age y Fare
qt_age = QuantileTransformer(output_distribution='normal', n_quantiles=500)
qt_fare = QuantileTransformer(output_distribution='normal', n_quantiles=500)
df['Age'] = qt_age.fit_transform(df[['Age']])
df['Fare'] = qt_fare.fit_transform(df[['Fare']])

# Escalar todas las columnas de 0 a 1 (menos Survived, que es lo que queremos predecir)
columnas_a_escalar = ['Pclass', 'Sex', 'Age', 'SibSp', 'Parch', 'Fare', 'Embarked']
for col in columnas_a_escalar:
    scaler = MinMaxScaler()
    df[col] = scaler.fit_transform(df[[col]])

df.sample(5)
```

```python
df.to_csv('titanic_procesado.csv', index=False)
print("Guardado correctamente")
```
