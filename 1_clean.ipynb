```python
import pandas as pd
df = pd.read_csv('titanic.csv')
df.sample(3)
```

```python
# Eliminar columna Cabin (le falta el 77% de los datos)
df.drop('Cabin', axis=1, inplace=True)

# Rellenar Age con la media
mean_age = df['Age'].mean()
df['Age'].fillna(mean_age, inplace=True)

# Rellenar Embarked con la moda
df['Embarked'] = df['Embarked'].fillna(df['Embarked'].mode()[0])

# Eliminar columnas que no aportan al modelo
df.drop('Name', axis=1, inplace=True)
df.drop('Ticket', axis=1, inplace=True)
df.drop('PassengerId', axis=1, inplace=True)

print(f"Después de limpiar df: {df.shape[0]} filas y {df.shape[1]} columnas.")
```

```python
df.to_csv('titanic_clean.csv', index=False)
print("Guardado correctamente")
```
